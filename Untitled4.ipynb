```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score)
# DATA LOAD
dt = pd.read_csv("customer_churn_dataset-training-master.csv")
dts = pd.read_csv("customer_churn_dataset-testing-master.csv")
dt.info()
dts.info()
dts.isnull().sum()
dt.isnull().sum()
dt[dt.isnull().any(axis=1)]
print(dt["Churn"].value_counts())
print("Training Shape:", dt.shape)
print("Testing Shape:", dts.shape)
# DATA CLEANING

dt = dt.dropna(how="all")
dts = dts.dropna(how="all")

print("\nTraining duplicates:", dt.duplicated().sum())
print("Testing duplicates:", dts.duplicated().sum())

dt = dt.drop_duplicates()
dts = dts.drop_duplicates()

print("\nTraining Shape After Cleaning:", dt.shape)
print("Testing Shape After Cleaning:", dts.shape)


# ADDED - CHECK MISSING VALUES AFTER CLEANING

print("\nMissing Values After Cleaning:")
print(dt.isnull().sum())


# ADDED - CHECK CATEGORICAL DATA

print("\nGender Categories:")
print(dt["Gender"].value_counts())

print("\nSubscription Type Categories:")
print(dt["Subscription Type"].value_counts())

print("\nContract Length Categories:")
print(dt["Contract Length"].value_counts())


# ADDED - DESCRIPTIVE STATISTICS

print("\nDescriptive Statistics:")
print(dt.describe())


# ADDED - STATISTICAL ANALYSIS BY CHURN

print("\nAverage Customer Characteristics by Churn:")

print(
    dt.groupby("Churn")[
        [
            "Age",
            "Tenure",
            "Usage Frequency",
            "Support Calls",
            "Payment Delay",
            "Total Spend",
            "Last Interaction"
        ]
    ].mean().round(2)
)


# EXPLORATORY DATA ANALYSIS (EDA)


# Churn Distribution

plt.figure(figsize=(6, 4))

sns.countplot(
    x="Churn",
    data=dt
)

plt.title("Churn Distribution")
plt.xlabel("Churn (0 = No, 1 = Yes)")
plt.ylabel("Number of Customers")

plt.show()


print("\nChurn Percentage:")

print(
    dt["Churn"].value_counts(normalize=True) * 100
)


# Gender vs Churn

plt.figure(figsize=(6, 4))

sns.countplot(
    x="Gender",
    hue="Churn",
    data=dt
)

plt.title("Gender vs Churn")

plt.show()


# Subscription Type vs Churn

plt.figure(figsize=(6, 4))

sns.countplot(
    x="Subscription Type",
    hue="Churn",
    data=dt
)

plt.title("Subscription Type vs Churn")

plt.show()


# Contract Length vs Churn

plt.figure(figsize=(6, 4))

sns.countplot(
    x="Contract Length",
    hue="Churn",
    data=dt
)

plt.title("Contract Length vs Churn")

plt.show()


# Age vs Churn

plt.figure(figsize=(6, 4))

sns.boxplot(
    x="Churn",
    y="Age",
    data=dt
)

plt.title("Age vs Churn")

plt.show()


# Tenure vs Churn

plt.figure(figsize=(6, 4))

sns.boxplot(
    x="Churn",
    y="Tenure",
    data=dt
)

plt.title("Tenure vs Churn")

plt.show()


# Support Calls vs Churn

plt.figure(figsize=(8, 4))

sns.countplot(
    x="Support Calls",
    hue="Churn",
    data=dt
)

plt.title("Support Calls vs Churn")

plt.show()


# Payment Delay vs Churn

plt.figure(figsize=(6, 4))

sns.boxplot(
    x="Churn",
    y="Payment Delay",
    data=dt
)

plt.title("Payment Delay vs Churn")

plt.show()


# Total Spend vs Churn

plt.figure(figsize=(6, 4))

sns.boxplot(
    x="Churn",
    y="Total Spend",
    data=dt
)

plt.title("Total Spend vs Churn")

plt.show()


# Histogram - Age Distribution

plt.figure(figsize=(7, 5))

sns.histplot(
    data=dt,
    x="Age",
    hue="Churn",
    bins=30,
    kde=True
)

plt.title("Age Distribution by Churn")

plt.show()


# ADDED - SCATTER PLOT

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=dt,
    x="Tenure",
    y="Total Spend",
    hue="Churn",
    alpha=0.5
)

plt.title("Tenure vs Total Spend")
plt.xlabel("Tenure")
plt.ylabel("Total Spend")

plt.show()


# Correlation Heatmap

numeric_dt = dt.select_dtypes(include="number")

plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_dt.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()


# ENCODING / PREPROCESSING

X_train = dt.drop(
    columns=["CustomerID", "Churn"]
)

y_train = dt["Churn"]


X_test = dts.drop(
    columns=["CustomerID", "Churn"]
)

y_test = dts["Churn"]


categorical_columns = [
    "Gender",
    "Subscription Type",
    "Contract Length"
]


X_train = pd.get_dummies(
    X_train,
    columns=categorical_columns,
    drop_first=True
)


X_test = pd.get_dummies(
    X_test,
    columns=categorical_columns,
    drop_first=True
)


X_train, X_test = X_train.align(
    X_test,
    join="left",
    axis=1,
    fill_value=0
)


print("\nEncoding completed")
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)


# FEATURE SCALING

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)


print("\nFeature scaling completed")


# MODEL TRAINING

logistic = LogisticRegression(
    max_iter=3000
)


decision_tree = DecisionTreeClassifier(
    random_state=42
)


random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)


knn = KNeighborsClassifier(
    n_neighbors=5
)


logistic.fit(
    X_train_scaled,
    y_train
)


knn.fit(
    X_train_scaled,
    y_train
)


decision_tree.fit(
    X_train,
    y_train
)


random_forest.fit(
    X_train,
    y_train
)


print("\nAll models trained successfully")


# MODEL EVALUATION

models = {

    "Logistic Regression":
        (logistic, X_test_scaled),

    "Decision Tree":
        (decision_tree, X_test),

    "Random Forest":
        (random_forest, X_test),

    "KNN":
        (knn, X_test_scaled)
}


results = []


for name, (model, test_data) in models.items():

    y_pred = model.predict(
        test_data
    )

    y_prob = model.predict_proba(
        test_data
    )[:, 1]


    accuracy = accuracy_score(
        y_test,
        y_pred
    )


    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )


    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )


    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )


    auc = roc_auc_score(
        y_test,
        y_prob
    )


    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1,
        auc
    ])


    # ADDED - CLASSIFICATION REPORT

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("\nClassification Report:")

    print(
        classification_report(
            y_test,
            y_pred,
            zero_division=0
        )
    )


    # CONFUSION MATRIX

    cm = confusion_matrix(
        y_test,
        y_pred
    )


    plt.figure(figsize=(5, 4))


    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues"
    )


    plt.title(
        name + " - Confusion Matrix"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")


    plt.show()


# MODEL COMPARISON

results_df = pd.DataFrame(

    results,

    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ]
)


print("\nModel Comparison:")

print(
    results_df.round(4)
)


# ADDED - MODEL COMPARISON GRAPH

comparison_plot = results_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],
    var_name="Metric",
    value_name="Score"
)


plt.figure(figsize=(11, 6))


sns.barplot(
    data=comparison_plot,
    x="Model",
    y="Score",
    hue="Metric"
)


plt.title("Machine Learning Model Comparison")

plt.xlabel("Model")
plt.ylabel("Score")

plt.ylim(0, 1)

plt.xticks(rotation=15)

plt.legend(
    title="Evaluation Metric"
)

plt.tight_layout()

plt.show()


# BEST MODEL SELECTION

results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)


print("\nModel Ranking:")

print(
    results_df.round(4)
)


best_model_name = results_df.loc[
    0,
    "Model"
]


print(
    "\nBest Model:",
    best_model_name
)


best_model = models[
    best_model_name
][0]


best_test_data = models[
    best_model_name
][1]


# CUSTOMER CHURN PREDICTION

churn_prediction = best_model.predict(
    best_test_data
)


churn_probability = best_model.predict_proba(
    best_test_data
)[:, 1]


prediction_result = pd.DataFrame({

    "CustomerID":
        dts["CustomerID"].values,

    "Actual Churn":
        y_test.values,

    "Predicted Churn":
        churn_prediction,

    "Churn Probability":
        churn_probability

})


print("\nCustomer Churn Predictions:")

print(
    prediction_result.head(10)
)


# RISK LEVEL

def get_risk_level(probability):

    if probability >= 0.70:

        return "High Risk"

    elif probability >= 0.40:

        return "Medium Risk"

    else:

        return "Low Risk"


prediction_result["Risk Level"] = (
    prediction_result[
        "Churn Probability"
    ].apply(get_risk_level)
)


print("\nCustomer Risk:")

print(
    prediction_result.head(10)
)


print("\nRisk Level Counts:")

print(
    prediction_result[
        "Risk Level"
    ].value_counts()
)


# RISK LEVEL VISUALIZATION

plt.figure(figsize=(7, 5))


sns.countplot(
    data=prediction_result,
    x="Risk Level",
    order=[
        "Low Risk",
        "Medium Risk",
        "High Risk"
    ]
)


plt.title(
    "Customer Churn Risk Levels"
)

plt.xlabel(
    "Risk Level"
)

plt.ylabel(
    "Number of Customers"
)


plt.show()


# FEATURE IMPORTANCE

feature_importance = pd.DataFrame({

    "Feature":
        X_train.columns,

    "Importance":
        random_forest.feature_importances_

})


feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)


print("\nTop Important Features:")

print(
    feature_importance.head(10)
)


# FEATURE IMPORTANCE VISUALIZATION

plt.figure(figsize=(10, 6))


sns.barplot(
    data=feature_importance.head(10),
    x="Importance",
    y="Feature"
)


plt.title(
    "Top 10 Features Affecting Customer Churn"
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)


plt.show()


# SAVE RESULTS

prediction_result.to_csv(
    "customer_churn_predictions.csv",
    index=False
)


results_df.to_csv(
    "model_comparison.csv",
    index=False
)


feature_importance.to_csv(
    "feature_importance.csv",
    index=False
)


print("\nAll results saved successfully")
```
